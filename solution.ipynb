## 3. Подготовка текста и восстановление заголовков

```python
%%writefile -a {PIPELINE}
HTML_TAG = re.compile(r'<[^>]+>')
SPACES = re.compile(r'\s+')

def clean_text(value):
    if pd.isna(value):
        return ''
    value = html.unescape(str(value))
    value = HTML_TAG.sub(' ', value)
    value = unicodedata.normalize('NFKC', value)
    value = value.replace('\u00ad', '')
    return SPACES.sub(' ', value).strip()

def prepare_texts(raw):
    out = raw.copy()
    for column in ['query', 'title', 'text']:
        out[column] = out[column].map(clean_text)
    empty_titles = {'—', '-', 'не указано', 'n/a', 'nan'}
    out.loc[out.title.str.casefold().isin(empty_titles), 'title'] = ''
    out['title'] = out.title.str.replace(
        r'\s*[-—–|]\s*Википедия\s*$', '', regex=True, case=False
    ).str.strip()
    for column in ['article_id', 'document_id']:
        missing = out[column].isna()
        out.loc[missing, column] = '__missing__' + out.loc[missing, 'id']
    return out.reset_index(drop=True)

train = prepare_texts(train_raw)
test = prepare_texts(test_raw)

def known_titles(frame):
    rows = frame.loc[frame.title.ne(''), ['article_id', 'title']]
    summary = rows.groupby('article_id').title.agg(['first', 'nunique'])
    return summary.loc[summary['nunique'].eq(1), 'first']

def restore_titles(frame, reference):
    out = frame.copy()
    combined = pd.concat([
        reference[['article_id', 'title']], frame[['article_id', 'title']]
    ], ignore_index=True)
    fill = out.article_id.map(known_titles(combined))
    missing = out.title.eq('') & fill.notna()
    out.loc[missing, 'title'] = fill[missing]
    return out

```

### NDCG@10

```python
%%writefile -a {PIPELINE}
def ndcg_per_query(frame, scores, k=10):
    scores = np.asarray(scores, dtype=np.float64)
    assert scores.shape == (len(frame),) and np.isfinite(scores).all()
    labels = frame.relevance.to_numpy(dtype=np.float64)
    ids = frame.id.astype(str).to_numpy()
    result = {}
    for qid, positions in frame.groupby('query_id', sort=False).indices.items():
        y = labels[positions]
        order = np.lexsort((ids[positions], -scores[positions]))[:k]
        discount = 1 / np.log2(np.arange(min(k, len(positions))) + 2)
        dcg = np.dot(np.exp2(y[order]) - 1, discount)
        ideal = np.sort(y)[::-1][:k]
        idcg = np.dot(np.exp2(ideal) - 1, discount)
        result[qid] = float(dcg / idcg) if idcg > 0 else 0.0
    return pd.Series(result, dtype=float)

def paired_interval(candidate, reference):
    delta = (candidate - reference.reindex(candidate.index)).to_numpy()
    assert np.isfinite(delta).all()
    rng = np.random.default_rng(CFG.seed)
    boot = rng.choice(delta, size=(2000, len(delta)), replace=True).mean(axis=1)
    return {
        'delta': float(delta.mean()),
        'ci95': np.quantile(boot, [0.025, 0.975]).tolist(),
        'queries': len(delta),
    }

```

## 5. Модель и токенизация

```python
%%writefile -a {PIPELINE}
tokenizer = AutoTokenizer.from_pretrained(
    CFG.model_name, revision=CFG.revision, use_fast=True
)
collator = DataCollatorWithPadding(tokenizer, pad_to_multiple_of=8)

def encode_frame(frame):
    queries = frame['query'].tolist()
    query_tokens = tokenizer(queries, add_special_tokens=False)['input_ids']
    queries = [
        text if len(tokens) <= 128 else tokenizer.decode(tokens[:128])
        for text, tokens in zip(queries, query_tokens)
    ]
    documents = [
        f'{title}\n{text}' if title else text
        for title, text in zip(frame.title, frame.text)
    ]
    encoded = tokenizer(
        queries, documents, truncation='only_second',
        max_length=CFG.max_length, padding=False,
    )
    return [
        {key: values[i] for key, values in encoded.items()}
        for i in range(len(frame))
    ]

```

```python
%%writefile -a {PIPELINE}
def new_model():
    seed_everything()
    config = AutoConfig.from_pretrained(
        CFG.model_name, revision=CFG.revision, trust_remote_code=True
    )
    config.use_flash_attn = False
    config.use_cache = False
    model, loading = AutoModelForSequenceClassification.from_pretrained(
        CFG.model_name, revision=CFG.revision, code_revision=CFG.revision,
        config=config, trust_remote_code=True, torch_dtype=torch.float32,
        output_loading_info=True,
    )
    assert not loading['missing_keys'] and not loading['unexpected_keys'], loading
    model.requires_grad_(False)
    for layer in model.roberta.encoder.layers[-CFG.trainable_layers:]:
        layer.requires_grad_(True)
    model.classifier.requires_grad_(True)
    model.roberta.encoder.gradient_checkpointing = True
    model.to(DEVICE)
    count = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print('Обучаемых параметров:', f'{count:,}')
    return model

def save_tuned_weights(model, path):
    state = {
        name: parameter.detach().cpu().clone()
        for name, parameter in model.named_parameters() if parameter.requires_grad
    }
    torch.save(state, path)

def load_tuned_weights(model, path):
    state = torch.load(path, map_location='cpu', weights_only=True)
    expected = {name for name, p in model.named_parameters() if p.requires_grad}
    assert set(state) == expected
    loaded = model.load_state_dict(state, strict=False)
    assert not loaded.unexpected_keys

def sha256_file(path):
    digest = hashlib.sha256()
    with open(path, 'rb') as source:
        for chunk in iter(lambda: source.read(1024 * 1024), b''):
            digest.update(chunk)
    return digest.hexdigest()

```

### Инференс

```python
%%writefile -a {PIPELINE}
def forward_scores(model, items):
    batch = collator(items)
    batch = {key: value.to(DEVICE) for key, value in batch.items()}
    with torch.autocast(
        DEVICE.type, dtype=torch.float16, enabled=DEVICE.type == 'cuda'
    ):
        scores = model(**batch, return_dict=True).logits.reshape(-1)
    return scores.float()

def predict(model, encoded, description):
    model.eval()
    order = np.argsort([len(row['input_ids']) for row in encoded], kind='stable')
    result = np.empty(len(encoded), dtype=np.float64)
    with torch.inference_mode():
        for start in tqdm(range(0, len(order), CFG.eval_batch), desc=description):
            positions = order[start:start + CFG.eval_batch]
            items = [encoded[i] for i in positions]
            result[positions] = forward_scores(model, items).cpu().numpy()
    assert np.isfinite(result).all()
    return result

```

## 6. Дообучение

```python
%%writefile -a {PIPELINE}
def ranking_loss(scores, labels):
    positive = torch.where(labels == 1)[0]
    negative = torch.where(labels == 0)[0]
    assert len(positive) and len(negative)
    differences = scores[positive, None] - scores[None, negative]
    pair_loss = F.softplus(-differences)

    order = torch.argsort(scores.detach(), descending=True, stable=True)
    ranks = torch.empty_like(order)
    ranks[order] = torch.arange(len(order), device=DEVICE)
    discounts = 1 / torch.log2(
        torch.arange(len(order), device=DEVICE, dtype=torch.float32) + 2
    )
    discounts[10:] = 0
    ideal = discounts[:min(len(positive), 10)].sum().clamp_min(1e-8)
    weights = (
        discounts[ranks[positive, None]] - discounts[ranks[None, negative]]
    ).abs() / ideal
    weighted = (pair_loss * weights).sum() / weights.sum().clamp_min(1e-8)
    binary = 0.5 * (
        F.softplus(-scores[positive]).mean() + F.softplus(scores[negative]).mean()
    )
    return 0.70 * weighted + 0.20 * pair_loss.mean() + 0.10 * binary

def training_groups(frame):
    groups = []
    for positions in frame.groupby('query_id', sort=False).indices.values():
        if frame.iloc[positions].relevance.nunique() < 2:
            continue
        ids = frame.iloc[positions].id.to_numpy()
        groups.append(positions[np.argsort(ids, kind='stable')])
    assert groups, 'Нет запросов с обоими классами.'
    return groups

def scores_for_query(model, encoded, positions):
    order = np.argsort([len(encoded[i]['input_ids']) for i in positions], kind='stable')
    sorted_positions = positions[order]
    pieces = []
    for start in range(0, len(positions), CFG.candidate_batch):
        batch_positions = sorted_positions[start:start + CFG.candidate_batch]
        pieces.append(forward_scores(model, [encoded[i] for i in batch_positions]))
    inverse = torch.as_tensor(np.argsort(order), device=DEVICE)
    return torch.cat(pieces)[inverse]

```

```python
%%writefile -a {PIPELINE}
def make_optimizer(model):
    head = []
    backbone = []
    for name, parameter in model.named_parameters():
        if not parameter.requires_grad:
            continue
        if name.startswith('classifier.'):
            head.append(parameter)
        else:
            backbone.append(parameter)
    return torch.optim.AdamW([
        {'params': backbone, 'lr': CFG.learning_rate},
        {'params': head, 'lr': CFG.head_learning_rate},
    ], weight_decay=CFG.weight_decay)

def update_weights(model, optimizer, scheduler, scaler):
    scaler.unscale_(optimizer)
    parameters = [p for p in model.parameters() if p.requires_grad]
    torch.nn.utils.clip_grad_norm_(parameters, 1.0)
    previous_scale = scaler.get_scale()
    scaler.step(optimizer)
    scaler.update()
    if scaler.get_scale() >= previous_scale:
        scheduler.step()
    optimizer.zero_grad(set_to_none=True)

```

```python
%%writefile -a {PIPELINE}
def train_block(model, frame, encoded, groups, order, optimizer, scheduler, scaler):
    model.train()
    optimizer.zero_grad(set_to_none=True)
    labels = frame.relevance.to_numpy(dtype=np.float32)
    losses = []
    for step, group_number in enumerate(tqdm(order, desc='Обучение Jina')):
        positions = groups[group_number]
        scores = scores_for_query(model, encoded, positions)
        target = torch.as_tensor(labels[positions], device=DEVICE)
        loss = ranking_loss(scores, target)
        if not torch.isfinite(loss):
            raise FloatingPointError('Неконечное значение функции потерь.')

        window_start = step // CFG.accumulate_queries * CFG.accumulate_queries
        window_size = min(CFG.accumulate_queries, len(order) - window_start)
        scaler.scale(loss / window_size).backward()
        losses.append(float(loss.detach().cpu()))
        is_update = (step + 1) % CFG.accumulate_queries == 0
        if is_update or step + 1 == len(order):
            update_weights(model, optimizer, scheduler, scaler)
    return float(np.mean(losses))

def fit_model(frame, encoded, model=None, validation=None, callback=None, fraction=1.0):
    if model is None:
        model = new_model()
    groups = training_groups(frame)
    order = np.random.default_rng(CFG.seed).permutation(len(groups))
    steps = math.ceil(len(groups) / CFG.accumulate_queries)
    optimizer = make_optimizer(model)
    scheduler = get_linear_schedule_with_warmup(
        optimizer, num_warmup_steps=max(1, round(steps * CFG.warmup_ratio)),
        num_training_steps=steps,
    )
    scaler = torch.amp.GradScaler(DEVICE.type, enabled=DEVICE.type == 'cuda')
    limit = math.ceil(fraction * len(groups) / CFG.accumulate_queries)
    limit = min(len(groups), max(1, limit * CFG.accumulate_queries))
    half = math.ceil(len(groups) / (2 * CFG.accumulate_queries)) * CFG.accumulate_queries
    boundaries = sorted({min(half, limit), limit})
    history = []
    start = 0
    for end in boundaries:
        loss = train_block(
            model, frame, encoded, groups, order[start:end], optimizer, scheduler, scaler
        )
        row = {'epoch_fraction': end / len(groups), 'loss': loss, 'queries': end}
        if validation is not None:
            scores = predict(model, validation, 'Validation Jina')
            callback(model, scores, row)
        history.append(row)
        print(row)
        start = end
    return model, pd.DataFrame(history)

```

### Выбор точки остановки и контроль качества

```python
%%writefile -a {PIPELINE}
selection = {'ndcg10': -np.inf, 'epoch_fraction': None}
best_weights = OUTPUT / 'best_validation_layers.pt'
best_scores = OUTPUT / 'best_validation_scores.npy'

def select_checkpoint(model, scores, row):
    score = float(ndcg_per_query(select_frame, scores[select_idx]).mean())
    row['selection_ndcg10'] = score
    if score > selection['ndcg10'] + 1e-12:
        selection.update(ndcg10=score, epoch_fraction=float(row['epoch_fraction']))
        save_tuned_weights(model, best_weights)
        np.save(best_scores, scores)

fit_encoded = encode_frame(fit_frame)
val_encoded = encode_frame(val_frame)
validation_model = new_model()
pretrained_scores = predict(validation_model, val_encoded, 'Jina до обучения')
validation_model, history = fit_model(
    fit_frame, fit_encoded, model=validation_model,
    validation=val_encoded, callback=select_checkpoint,
)
load_tuned_weights(validation_model, best_weights)
validation_scores = np.load(best_scores)
history.to_csv(OUTPUT / 'training_history.csv', index=False)
print('Выбрана точка:', selection)

audit_tuned = ndcg_per_query(audit_frame, validation_scores[audit_idx])
audit_pretrained = ndcg_per_query(audit_frame, pretrained_scores[audit_idx])
report = {
    'model': CFG.model_name,
    'selection': selection,
    'validation_ndcg10': float(ndcg_per_query(val_frame, validation_scores).mean()),
    'audit_ndcg10': float(audit_tuned.mean()),
    'audit_pretrained_ndcg10': float(audit_pretrained.mean()),
    'audit_tuned_minus_pretrained': paired_interval(audit_tuned, audit_pretrained),
}
(OUTPUT / 'validation_report.json').write_text(
    json.dumps(report, ensure_ascii=False, indent=2), encoding='utf-8'
)
validation_output = val_frame[['id', 'query_id', 'relevance']].copy()
validation_output['prediction'] = validation_scores
validation_output['pretrained_prediction'] = pretrained_scores
validation_output['is_audit'] = audit_mask
validation_output.to_csv(OUTPUT / 'validation_predictions.csv', index=False)

print(f"Validation NDCG@10: {report['validation_ndcg10']:.6f} | Контроль: {report['audit_ndcg10']:.6f}")

del validation_model, fit_encoded, val_encoded
gc.collect()
torch.cuda.empty_cache()

```
